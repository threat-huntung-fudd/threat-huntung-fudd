```python
%load_ext autoreload
%autoreload 2
import sys
from pathlib import Path
import logging
import os
import logging
import gdown
import tarfile
# Set system path to the folder where the notebook is saved
notebook_dir = Path().resolve()
sys.path.insert(0, str(notebook_dir))

from cadets import DarpaParser

# Set up logging configuration
logging.basicConfig(
    level=logging.INFO, 
    format="%(asctime)s - %(levelname)s - %(message)s",
    handlers=[
        logging.StreamHandler(sys.stdout)  # Ensure logs are output to stdout
    ]
)
logger = logging.getLogger(__name__)
```

```python
def download_data(urls: list[str], save_dir: str):
    logging.info("Executing DownloadData")
    os.makedirs(save_dir, exist_ok=True)  # Ensure save directory exists
    
    for url in urls:
        logging.info(f"Downloading {url}")
        filename = gdown.download(url, quiet=False, use_cookies=False, fuzzy=True)
        
        if filename:
            destination = os.path.join(save_dir, os.path.basename(filename))
            os.rename(filename, destination)
        else:
            logging.warning(f"Failed to download {url}")
```
